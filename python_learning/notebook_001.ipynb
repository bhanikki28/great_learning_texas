# Python Fundamentals: Built-in Data Types

A data type describes what kind of value a variable holds and what you can do with it. Python has several built-in types. Run the code cells as you read them; `type(value)` shows a value's type.

## 1. Numbers: `int`, `float`, and `complex`

Use `int` for whole numbers, `float` for decimal numbers, and `complex` for numbers with a real and imaginary part. Python automatically chooses a numeric type based on the value.

```python
age = 25                 # int
price = 19.95            # float
measurement = 2 + 3j     # complex

print(age, type(age))
print(price, type(price))
print(measurement, type(measurement))
print(age + 1)            # arithmetic works with numbers
```

## 2. Text: `str`

A string stores text. Put text inside single or double quotes. Strings are immutable, so operations such as `.upper()` return a new string rather than changing the original.

```python
name = 'Ada Lovelace'
greeting = 'Hello, ' + name
print(greeting)
print(name.upper())
print(type(name))
```

## 3. Boolean: `bool`

A Boolean value is either `True` or `False`. Booleans are often used in comparisons and `if` statements.

```python
temperature = 22
is_warm = temperature > 20
print(is_warm, type(is_warm))

if is_warm:
    print('It is warm today.')
```

## 4. Sequences: `list`, `tuple`, and `range`

Sequences keep items in order. A `list` can be changed, a `tuple` cannot be changed after creation, and a `range` represents a sequence of whole numbers (often used in loops). Items are accessed by position, starting at index `0`.

```python
employee_list = ['Sachin', 'Poovalingam', 'Saroja', 'Bharathy']
employee_list.append('Maya')  # lists can be changed
print(employee_list)
print('First employee:', employee_list[0])

coordinates = (10, 20)        # tuple: fixed pair of values
print(coordinates, type(coordinates))

numbers = range(1, 5)          # 1, 2, 3, 4
print(list(numbers))
```

## 5. Set: `set`

A set stores unique items and does not guarantee their order. It is useful for removing duplicates and checking membership.

```python
colors = {'blue', 'green', 'blue'}
print(colors)              # 'blue' appears only once
print('green' in colors)
```

## 6. Dictionary: `dict`

A dictionary stores values under keys. Use a key to look up its associated value.

```python
employee = {'name': 'Saroja', 'department': 'Design', 'years': 4}
print(employee['name'])
employee['years'] = 5        # dictionaries can be changed
print(employee)
```

## 7. No value: `NoneType`

`None` means that a value is absent or not set. Its type is `NoneType`.

```python
middle_name = None
print(middle_name, type(middle_name))

if middle_name is None:
    print('No middle name was provided.')
```

## 8. Bytes: `bytes` and `bytearray`

These types hold sequences of byte values, commonly used for binary data such as files or network content. `bytes` is immutable; `bytearray` can be changed. Most beginner programs work mainly with strings instead.

```python
message_bytes = b'Hi'
editable_bytes = bytearray(b'Hi')
editable_bytes[0] = 72  # 72 is the byte value for 'H'
print(message_bytes, type(message_bytes))
print(editable_bytes, type(editable_bytes))
```

## Choosing a type

| Need | Type | Example |
|---|---|---|
| A whole or decimal number | `int`, `float` | `42`, `3.14` |
| Text | `str` | `'hello'` |
| True/false result | `bool` | `is_ready = True` |
| Ordered items you may edit | `list` | `[1, 2, 3]` |
| Ordered items you keep fixed | `tuple` | `(1, 2, 3)` |
| Unique items | `set` | `{'red', 'blue'}` |
| Key/value lookup | `dict` | `{'name': 'Ada'}` |
| No value yet | `NoneType` | `result = None` |

Convert compatible values with constructors such as `int('12')`, `float('2.5')`, and `str(12)`. For example, `int('12') + 3` evaluates to `15`. Not every conversion is valid: `int('hello')` raises a `ValueError`.
